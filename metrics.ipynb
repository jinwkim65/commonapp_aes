```python
import re

def calculate_ps_ratio(text):
    # Extract sections
    problem_sections = re.findall(r'\[P\](.*?)\[/P\]', text, re.DOTALL)
    solution_sections = re.findall(r'\[S\](.*?)\[/S\]', text, re.DOTALL)
    takeaway_sections = re.findall(r'\[T\](.*?)\[/T\]', text, re.DOTALL)

    # Count words in each section
    problem_words = sum(len(section.split()) for section in problem_sections)
    solution_words = sum(len(section.split()) for section in solution_sections)
    takeaway_words = sum(len(section.split()) for section in takeaway_sections)

    total_words = problem_words + solution_words + takeaway_words

    # Handle edge case where total_words is 0
    if total_words == 0:
        return "No words found in the input."

    # Calculate percentages
    problem_pct = (problem_words / total_words) * 100
    solution_pct = (solution_words / total_words) * 100
    takeaway_pct = (takeaway_words / total_words) * 100

    # Ideal ratios
    ideal = {"Problem": 20, "Solution": 60, "Takeaway": 20}

    # Calculate deviation from ideal
    deviation = abs(problem_pct - ideal['Problem']) + abs(solution_pct - ideal['Solution']) + abs(takeaway_pct - ideal['Takeaway'])
    ps_ratio_score = max(0, 10 - (deviation / 100) * 10)

    # Return results
    return {
        "Problem Words": problem_words,
        "Solution Words": solution_words,
        "Takeaway Words": takeaway_words,
        "Total Words": total_words,
        "Percentages": {
            "Problem %": problem_pct,
            "Solution %": solution_pct,
            "Takeaway %": takeaway_pct,
        },
        "PS Ratio Score": round(ps_ratio_score, 2)
    }
```
