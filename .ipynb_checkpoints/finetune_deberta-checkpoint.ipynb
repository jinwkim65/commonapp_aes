Part I

```python
# Import libraries
from transformers import AutoTokenizer
from collections import defaultdict
import os

def read_text_files(*filenames):
    """
    Reads multiple text files and returns their contents as a list.
    
    Args:
        *filenames: Variable number of text file names (strings).
    
    Returns:
        List of file contents, where each element corresponds to a file.
    """
    file_contents = []
    
    for filename in filenames:
        try:
            with open(filename, "r", encoding="utf-8") as file:
                file_contents.append(file.read())
        except FileNotFoundError:
            print(f"Warning: {filename} not found.")
            file_contents.append(None)
        except Exception as e:
            print(f"Error reading {filename}: {e}")
            file_contents.append(None)
    
    return file_contents

# Load text files from data folder into corpus
corpus = read_text_files("legal_document.txt", "lit_work.txt")

# Pre-tokenizer
tokenizer = AutoTokenizer.from_pretrained("bert-base-cased")
```

```python
# Compute frequencies of each word in corpus as we do pre-tokenization

word_freqs = defaultdict(int)
for text in corpus:
    words_with_offsets = tokenizer.backend_tokenizer.pre_tokenizer.pre_tokenize_str(text)
    new_words = [word for word, offset in words_with_offsets]
    for word in new_words:
        word_freqs[word] += 1

# Display frequencies of the first 10 words
print(dict(list(word_freqs.items())[:10]))
```

```python
# Generate alphabet
alphabet = []
for word in word_freqs.keys():
    if word[0] not in alphabet:
        alphabet.append(word[0])
    for letter in word[1:]:
        if f"##{letter}" not in alphabet:
            alphabet.append(f"##{letter}")

alphabet.sort()
alphabet
```

```python
# Special tokens used by the model (BERT)
vocab = ["[PAD]", "[UNK]", "[CLS]", "[SEP]", "[MASK]"] + alphabet.copy()

# Split each word not prefixed by '##'
splits = {
    word: [c if i == 0 else f"##{c}" for i, c in enumerate(word)]
    for word in word_freqs.keys()
}

# Function to computes the score of each pair
def compute_pair_scores(splits):
    letter_freqs = defaultdict(int)
    pair_freqs = defaultdict(int)
    for word, freq in word_freqs.items():
        split = splits[word]
        if len(split) == 1:
            letter_freqs[split[0]] += freq
            continue
        for i in range(len(split) - 1):
            pair = (split[i], split[i + 1])
            letter_freqs[split[i]] += freq
            pair_freqs[pair] += freq
        letter_freqs[split[-1]] += freq

    scores = {
        pair: freq / (letter_freqs[pair[0]] * letter_freqs[pair[1]])
        for pair, freq in pair_freqs.items()
    }
    return scores

# Compute pair scores on our corpus
pair_scores = compute_pair_scores(splits)

# Display the top 5 pair scores
for i, key in enumerate(pair_scores.keys()):
    print(f"{key}: {pair_scores[key]}")
    if i >= 5:
        break
```

```python
# Function to merge a pair of tokens
def merge_pair(a, b, splits):
    for word in word_freqs:
        split = splits[word]
        if len(split) == 1:
            continue
        i = 0
        while i < len(split) - 1:
            if split[i] == a and split[i + 1] == b:
                merge = a + b[2:] if b.startswith("##") else a + b
                split = split[:i] + [merge] + split[i + 2 :]
            else:
                i += 1
        splits[word] = split
    return splits

# Merge to a fixed vocabulary size (250)
vocab_size = 250
while len(vocab) < vocab_size:
    scores = compute_pair_scores(splits)
    best_pair, max_score = "", None
    for pair, score in scores.items():
        if max_score is None or max_score < score:
            best_pair = pair
            max_score = score
    splits = merge_pair(*best_pair, splits)
    new_token = (
        best_pair[0] + best_pair[1][2:]
        if best_pair[1].startswith("##")
        else best_pair[0] + best_pair[1]
    )
    vocab.append(new_token)

# Display the final vocabulary
print(vocab)
```

```python
# Function to encode a word into tokens from vocabulary
def encode_word(word):
    tokens = []
    while len(word) > 0:
        i = len(word)
        while i > 0 and word[:i] not in vocab:
            i -= 1
        if i == 0:
            return ["[UNK]"]
        tokens.append(word[:i])
        word = word[i:]
        if len(word) > 0:
            word = f"##{word}"
    return tokens

# Function to tokenize text
def tokenize(text):
    pre_tokenize_result = tokenizer._tokenizer.pre_tokenizer.pre_tokenize_str(text)
    pre_tokenized_text = [word for word, offset in pre_tokenize_result]
    encoded_words = [encode_word(word) for word in pre_tokenized_text]
    return sum(encoded_words, [])
```

```python
# Tokenize legal document
with open("./legal_document.txt", "r", encoding="utf-8") as f:
    legal_text = f.read()

legal_tokens = tokenize(legal_text)

# Tokenize literary work
with open("./lit_work.txt", "r", encoding="utf-8") as f:
    lit_work = f.read()

lit_tokens = tokenize(lit_work)
```

PART I

```python
import torch
import torch.nn as nn
import torch.optim as optim

# Check for GPU availability
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Using device: {device}")
```

```python
import gensim
import gensim.models
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity
from collections import defaultdict

# Gensim uses lists of lists, so convert tokenized text accordingly
legal_tokens = [legal_tokens]
lit_tokens = [lit_tokens]

# Train Word2Vec (CBOW and Skip-gram) on Legal Document
cbow_legal = gensim.models.Word2Vec(legal_tokens, vector_size=50, window=5, min_count=1, sg=0)
skipgram_legal = gensim.models.Word2Vec(legal_tokens, vector_size=50, window=5, min_count=1, sg=1)

# Train Word2Vec (CBOW and Skip-gram) on Literary Work
cbow_literary = gensim.models.Word2Vec(lit_tokens, vector_size=50, window=5, min_count=1, sg=0)
skipgram_literary = gensim.models.Word2Vec(lit_tokens, vector_size=50, window=5, min_count=1, sg=1)

# Build Vocabulary
cbow_legal.build_vocab(legal_tokens)
skipgram_legal.build_vocab(legal_tokens)

cbow_literary.build_vocab(lit_tokens)
skipgram_literary.build_vocab(lit_tokens)

# Train Models
cbow_legal.train(legal_tokens, total_examples=len(legal_tokens), epochs=10)
skipgram_legal.train(legal_tokens, total_examples=len(legal_tokens), epochs=10)

cbow_literary.train(lit_tokens, total_examples=len(lit_tokens), epochs=10)
skipgram_literary.train(lit_tokens, total_examples=len(lit_tokens), epochs=10)
```

```python
# Check Vocabulary for Legal Text
print("CBOW Model Vocabulary for Legal Text:", cbow_legal.wv.index_to_key[:100])  # Show first 100 tokens
print("Skip-gram Model Vocabulary for Legal Text:", skipgram_legal.wv.index_to_key[:10])

# Check Vocabulary for Literary Text
print("CBOW Model Vocabulary for Literary Text:", cbow_literary.wv.index_to_key[:10])  # Show first 10 tokens
print("Skip-gram Model Vocabulary for Literary Text:", skipgram_literary.wv.index_to_key[:10])
```

```python
# 1. Implement a procedure that computes the cosine similarity between two words (vectors).
from sklearn.metrics.pairwise import cosine_similarity

def compute_cosine_similarity(word1, word2, model):
    """
    Computes cosine similarity between two words in a trained Word2Vec model.

    Parameters:
        word1 (str): First word or subword token.
        word2 (str): Second word or subword token.
        model (gensim.models.Word2Vec): Trained Word2Vec model.

    Returns:
        float: Cosine similarity score between -1 and 1.
    """
    if word1 in model.wv and word2 in model.wv:
        vec1 = model.wv[word1].reshape(1, -1)  # Reshape to work with sklearn
        vec2 = model.wv[word2].reshape(1, -1)
        return cosine_similarity(vec1, vec2)[0][0]
    else:
        return None  # Return None if a word is not found in the vocabulary

# Testing cosine similarity for some word pairs

word_pairs = [
    ("##t", "##e"),
    ("##o", "##a"),
    ("##9", "_")
]

for w1, w2 in word_pairs:
    similarity = compute_cosine_similarity(w1, w2, cbow_legal)
    if similarity is not None:
        print(f"Cosine Similarity between '{w1}' and '{w2}': {similarity:.4f}")
    else:
        print(f"One of the words '{w1}' or '{w2}' is not in the vocabulary.")
```

```python
# 2. Find pairs of words that are very similar in one encoding and not very similar in the other encoding.

# Get common words between CBOW and Skip-gram models
vocab = list(set(cbow_legal.wv.index_to_key) & set(skipgram_legal.wv.index_to_key))

# Find word pairs with opposing similarities
matching_pairs = []

# Nested loops to compare all possible word pairs
for i in range(len(vocab)):
    for j in range(i + 1, len(vocab)):  # Avoid duplicate comparisons
        word1 = vocab[i]
        word2 = vocab[j]

        sim_cbow = compute_cosine_similarity(word1, word2, cbow_legal)
        sim_skipgram = compute_cosine_similarity(word1, word2, skipgram_legal)

        if sim_cbow is not None and sim_skipgram is not None:
            # Check if they meet the condition: similar in CBOW, dissimilar in Skip-gram
            if (0 < sim_cbow < 1) and (-1 < sim_skipgram < 0):
                matching_pairs.append((word1, word2, sim_cbow, sim_skipgram))

# Sort by difference in similarity
matching_pairs.sort(key=lambda x: abs(x[2] - x[3]), reverse=True)

# Print top 5 most extreme pairs
for w1, w2, sim_cb, sim_sg in matching_pairs[:5]:
    print(f"Pair: ('{w1}', '{w2}')")
    print(f"  CBOW Similarity: {sim_cb:.4f}")
    print(f"  Skip-gram Similarity: {sim_sg:.4f}")
    print(f"  Difference: {abs(sim_cb - sim_sg):.4f}\n")
```

```python
# 3. Find the pair of vectors that have the largest disparity.

# Get common words between CBOW and Skip-gram models
vocab = list(set(cbow_legal.wv.index_to_key) & set(skipgram_legal.wv.index_to_key))

# Variables to track the pair with the largest disparity
max_disparity = 0
max_pair = None

# Nested loops to check all word pairs
for i in range(len(vocab)):
    for j in range(i + 1, len(vocab)):  # Avoid duplicate pairs
        word1 = vocab[i]
        word2 = vocab[j]

        sim_cb = compute_cosine_similarity(word1, word2, cbow_legal)
        sim_sg = compute_cosine_similarity(word1, word2, skipgram_legal)

        if sim_cb is not None and sim_sg is not None:
            disparity = abs(sim_cb - sim_sg)  # Compute the absolute difference
            if disparity > max_disparity:
                max_disparity = disparity
                max_pair = (word1, word2, sim_cb, sim_sg)

# Print the pair with the largest disparity
if max_pair:
    w1, w2, sim_cb, sim_sg = max_pair
    print(f"Pair with largest disparity: ('{w1}', '{w2}')")
    print(f"  CBOW Similarity: {sim_cb:.4f}")
    print(f"  Skip-gram Similarity: {sim_sg:.4f}")
    print(f"  Absolute Disparity: {max_disparity:.4f}")
```

Part II

```python
import re

# Read combined text file
text = read_text_files("combined-text.txt")[0]

# Split cases based on 5 or more newlines
cases = re.split(r'\n{5,}', text.strip())

# Dictionary to store case names and their context
case_dict = {}

# Regex to capture full "X v. Y" name, stopping at the first number
case_name_pattern = re.compile(r'\b[A-Z][a-zA-Z]+ v\. (?:[A-Z][a-zA-Z]+(?:[-+&’\' ][A-Za-z]+)* )*[^0-9]+')

# Process extracted case sections
for case_text in cases:
    lines = case_text.strip().split("\n")  # Split into lines
    if len(lines) > 1:  # Ensure there's both a title and content
        case_title = lines[0].strip()  # First line is the case name
        
        # Extract full "X v. Y" case name until the first number
        match = case_name_pattern.search(case_title)
        if match:
            short_case_name = match.group().strip()  # Extract and clean case name
            short_case_name = short_case_name.rstrip(",")  # Remove trailing comma if present
            case_context = " ".join(lines[1:]).strip()  # The rest is context
            case_dict[short_case_name] = case_context  # Store in dictionary

# Print first 5 extracted cases for verification
print("\nFirst 5 Extracted Cases:\n")
for i, (case, context) in enumerate(case_dict.items()):
    if i == 5:
        break
    print(f"{i+1}. Case: {case}\n   Context: {context[:200]}...\n")

# Print total number of cases extracted
print(f"\nTotal Cases Extracted: {len(case_dict)}")
```

```python
import re
import nltk

# Download necessary NLTK models if needed
nltk.download("punkt")

# Read the LLM response file
file_path = "Model-output.txt"
with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

# Remove prompt from file
legal_start_match = re.search(r'\bI\.\s', text)
if legal_start_match:
    text = text[legal_start_match.start():]

# Remove all quotation marks (both straight and curly quotes)
text = re.sub(r'[“”‘’"]', '', text)

# Normalize whitespace
text = re.sub(r'\s+', ' ', text)  # Replace multiple spaces/newlines with a single space

# Tokenize text into sentences
sentences = nltk.sent_tokenize(text)

# Regex pattern to capture case names
case_pattern = re.compile(
    r"\b[A-Z][a-zA-Z]+ v\. (?:[A-Z][a-zA-Z]+(?:[-'’]?[A-Za-z]+)*(?: [A-Z][a-zA-Z]+(?:[-'’]?[A-Za-z]+)*)*)"
)

# Dictionary to store case citations and their correct preceding sentence
case_citations = {}

# Iterate through sentences and find citations
for i in range(1, len(sentences)):  # Start from 1 to ensure there's a previous sentence
    matches = case_pattern.findall(sentences[i])  # Find case citations in current sentence

    if matches:  # If citations are found
        # Step 1: Identify the correct preceding sentence
        valid_preceding_sentence = sentences[i - 1].strip()

        # Step 2: If the preceding sentence contains a citation, move back further
        while case_pattern.search(valid_preceding_sentence) and i > 1:
            i -= 1
            valid_preceding_sentence = sentences[i - 1].strip()

        # Step 3: Store the correct context for all citations in this sentence
        for match in matches:
            # Step 4: Remove citation if context contains "aff'd"
            if "aff'd" not in valid_preceding_sentence.lower():
                case_citations[match] = valid_preceding_sentence  # Assign correct legal discussion

# Print first 10 extracted case citations with correct context
print("\nFirst 10 Extracted Citations with Correct Context:\n")
for i, (case, context) in enumerate(case_citations.items()):
    if i == 10:
        break
    print(f"{i+1}. Citation: {case}\n   Context: {context}\n")

# Print total number of citations extracted
print(f"\nTotal Citations Extracted: {len(case_citations)}")
```

```python
import numpy as np
from sentence_transformers import SentenceTransformer

### Embeddings for cases from combined-text.txt

# Load a different embedding model (new model for Part 2)
model = SentenceTransformer('sentence-transformers/all-mpnet-base-v2')

# Real cases extracted from combined-text.txt (Dictionary: {case_title: case_text})
real_cases = case_dict  

# Create a dictionary to store embeddings
case_embeddings_dict = {}

# Embed case titles and contexts
for case_title, case_context in real_cases.items():
    title_embedding = model.encode(case_title, convert_to_tensor=False)  # Convert title to embedding
    context_embedding = model.encode(case_context, convert_to_tensor=False)  # Convert context to embedding

    # Store embeddings in dictionary
    case_embeddings_dict[case_title] = {
        "title_embedding": title_embedding,
        "context_embedding": context_embedding
    }

# Print confirmation
print(f"Stored embeddings for {len(case_embeddings_dict)} cases in a dictionary.")
```

```python
# Extracted citations from model-output.txt (Dictionary: {citation: sentence_context})
llm_citations = case_citations  

# Create a dictionary to store LLM citation embeddings
llm_embeddings_dict = {}

# Embed LLM case citations and contexts
for case_title, case_context in llm_citations.items():
    title_embedding = model.encode(case_title, convert_to_tensor=False)  # Convert title to embedding
    context_embedding = model.encode(case_context, convert_to_tensor=False)  # Convert context to embedding

    # Store embeddings in dictionary
    llm_embeddings_dict[case_title] = {
        "title_embedding": title_embedding,
        "context_embedding": context_embedding
    }

# Print confirmation
print(f"Stored embeddings for {len(llm_embeddings_dict)} LLM-cited cases in a dictionary.")
```

```python
import numpy as np
import re
from sklearn.metrics.pairwise import cosine_similarity

# Set similarity threshold for hallucination detection
SIMILARITY_THRESHOLD = 0.8

# Function to normalize case titles
def normalize_case_title(title):
    """Remove punctuation, convert to lowercase, and simplify names."""
    title = title.lower()  # Convert to lowercase
    title = re.sub(r'[^\w\s]', '', title)  # Remove punctuation
    return title.strip()

# Dictionary to store hallucinated cases
nonexistent_cases = {}

# Step 1: Check each LLM-cited case
for llm_case, llm_embeds in llm_embeddings_dict.items():
    if llm_case in case_embeddings_dict:  # Exact match check
        continue  # This case exists, so it's not a hallucination

    # Normalize LLM case title
    normalized_llm_case = normalize_case_title(llm_case)

    llm_title_emb = llm_embeds["title_embedding"].reshape(1, -1)

    best_match = None
    best_similarity = -1  # Initialize as lowest possible similarity

    for real_case, real_embeds in case_embeddings_dict.items():
        # Normalize real case title
        normalized_real_case = normalize_case_title(real_case)

        real_title_emb = real_embeds["title_embedding"].reshape(1, -1)

        # Compute title similarity only
        title_similarity = cosine_similarity(llm_title_emb, real_title_emb)[0][0]

        if title_similarity > best_similarity:
            best_similarity = title_similarity
            best_match = real_case  # Store the closest matching real case

    # Step 3: If the best similarity is below threshold, it's a hallucination
    if best_similarity < SIMILARITY_THRESHOLD:
        nonexistent_cases[llm_case] = {
            "LLM Context": llm_citations[llm_case],  # Where the citation appears
            "Closest Real Case": best_match,
            "Similarity": best_similarity
        }

# Print results
print(f"\n🚨 Type 1 (Nonexistent Case Citations): {len(nonexistent_cases)} found\n")
for i, (case, details) in enumerate(nonexistent_cases.items()):
    if i == 10:  # Show only the first 10 for readability
        break
    print(f"{i+1}. ❌ Citation: {case}\n   Context: {details['LLM Context'][:200]}...\n"
          f"   🔍 Closest Real Case: {details['Closest Real Case']} (Similarity: {details['Similarity']:.2f})\n")
```

```python
import numpy as np
import re
from sklearn.metrics.pairwise import cosine_similarity

# Set similarity thresholds
TITLE_SIMILARITY_THRESHOLD = 0.8
MISINTERPRETATION_THRESHOLD = 0.7

# Function to normalize case titles
def normalize_case_title(title):
    """Remove punctuation, convert to lowercase, and simplify names."""
    title = title.lower()  # Convert to lowercase
    title = re.sub(r'[^\w\s]', '', title)  # Remove punctuation
    return title.strip()

# Dictionary to store similarity scores and misinterpretations
case_similarity_scores = {}
misinterpreted_cases = {}

# Step 1: Check each LLM-cited case
for llm_case, llm_embeds in llm_embeddings_dict.items():
    llm_title_emb = llm_embeds["title_embedding"].reshape(1, -1)
    llm_context_emb = llm_embeds["context_embedding"].reshape(1, -1)

    best_match = None
    best_title_similarity = -1  # Initialize lowest similarity score

    # Step 2: Find the closest matching real case based on title similarity
    for real_case, real_embeds in case_embeddings_dict.items():
        real_title_emb = real_embeds["title_embedding"].reshape(1, -1)

        # Compute title similarity
        title_similarity = cosine_similarity(llm_title_emb, real_title_emb)[0][0]

        if title_similarity > best_title_similarity:
            best_title_similarity = title_similarity
            best_match = real_case  # Store the closest matching real case

    # Step 3: Only compare contexts if title similarity is above 0.8
    if best_match and best_title_similarity >= TITLE_SIMILARITY_THRESHOLD:
        real_context_emb = case_embeddings_dict[best_match]["context_embedding"].reshape(1, -1)
        context_similarity = cosine_similarity(llm_context_emb, real_context_emb)[0][0]

        # Store similarity results
        case_similarity_scores[llm_case] = {
            "Title Similarity": best_title_similarity,
            "Context Similarity": context_similarity,
            "Closest Real Case": best_match,
            "LLM Context": llm_citations[llm_case],  # LLM's discussion
            "Actual Case Text": case_dict[best_match]  # Real case text
        }

        # Step 4: Flag as misinterpreted if context similarity is too low
        if context_similarity < MISINTERPRETATION_THRESHOLD:
            misinterpreted_cases[llm_case] = case_similarity_scores[llm_case]

# Sort cases by title similarity (highest to lowest)
sorted_cases = sorted(case_similarity_scores.items(), key=lambda x: x[1]["Title Similarity"], reverse=True)

# Print results for all cases
print(f"\n📊 Type 2 (All Case Similarity Scores): {len(sorted_cases)} cases compared\n")
for i, (case, details) in enumerate(sorted_cases):
    if i == 20:  # Show only the first 20 for readability
        break
    print(f"{i+1}. 📖 Citation: {case}\n"
          f"   🔍 Title Similarity: {details['Title Similarity']:.2f}\n"
          f"   🔍 Context Similarity: {details['Context Similarity']:.2f}\n"
          f"   🔍 Closest Match: {details['Closest Real Case']}\n"
          f"   🔹 LLM's Discussion: {details['LLM Context'][:200]}...\n"
          f"   🔸 Actual Case Text: {details['Actual Case Text'][:200]}...\n")

# Print total number of cases analyzed
print(f"\nTotal Cases Analyzed: {len(sorted_cases)}")
```
