```python
import transformers
import torch

model_id = "meta-llama/Meta-Llama-3-8B-Instruct"

pipeline = transformers.pipeline(
    "text-generation",
    model=model_id,
    model_kwargs={"torch_dtype": torch.bfloat16},
    device_map="auto",
)

messages = [
    {"role": "system", "content": "You are an excessively pedantic college professor."},
    {"role": "user", "content": "Teach me something about quantum physics."},
]

terminators = [
    pipeline.tokenizer.eos_token_id,
    pipeline.tokenizer.convert_tokens_to_ids("<|eot_id|>")
]

outputs = pipeline(
    messages,
    max_new_tokens=256,
    eos_token_id=terminators,
    do_sample=True,
    temperature=0.6,
    top_p=0.9,
)
print(outputs[0]["generated_text"][-1])
```

```python
import time
import torch
import transformers

def evaluate_essay(essay, pipeline, terminators, weaknesses_dict): 

    print("Thanks so much for trusting us with your writing! Let's begin the analysis for your essay.")
    
    # Strengths
    evaluator_role = "You are an expert college admissions counselor. You are known for your encouraging, actionable, and specific feedback on college application essays."
    evaluator_prompt = f"Summarize the essay in 4-6 bullet points. Identify 2-3 core themes in the essay. Highlight 1-2 excerpts from the essay that are impactful. Essay: {essay}"
    
    aes_messages = [
        {"role": "system", "content": evaluator_role},
        {"role": "user", "content": evaluator_prompt}
    ]
    
    with torch.no_grad():
        aes_outputs = pipeline(
            aes_messages,
            max_new_tokens=512,
            eos_token_id=terminators,
            do_sample=True,
            temperature=0.6,
            top_p=0.9,
        )
    
    strengths = aes_outputs[0]["generated_text"][-1]["content"]
    print(strengths)

    # Weaknesses
    weakness_prompt = f"""
    Identify 2-3 of the most significant weaknesses which are present in the essay below, based on the following weaknesses:
    {', '.join([f'{weakness}: {description}' for weakness, description in weaknesses_dict.items()])}
    
    Output the weaknesses in the following format:
    Weakness 1: Definition of weakness 1
    Weakness 2: Definition of weakness 2
    Weakness 3: Definiton of weakness 3
    
    For example:
    Weak Conclusion: The conclusion lacks impact, often summarizing points without providing a final, strong reflection or takeaway. A more thoughtful and impactful closing could leave a lasting impression.
    Lacks Clear Theme: The essay feels unfocused, without a clear central theme or message. It may jump between topics without weaving them together under one cohesive narrative thread.
    Lack of Conclusion or Summary: The essay ends abruptly, with no clear conclusion or takeaway. The lack of closure can leave the reader feeling unsatisfied or unclear about the essay's central message.
    
    Do not output anything else.
    Essay: {sample_essay}
    """

    aes_messages = [
        {"role": "system", "content": "You are an expert in evaluating and providing feedback on college essays."},
        {"role": "user", "content": weakness_prompt}
    ]

    with torch.no_grad():
        aes_outputs = pipeline(
            aes_messages,
            max_new_tokens=256,
            eos_token_id=terminators,
            do_sample=True,
            temperature=0.6,
            top_p=0.9,
        )

    weaknesses = aes_outputs[0]["generated_text"][-1]["content"]

    # Evaluate based on extracted weaknesses
    full_weakness_prompt = f"""
    Evaluate the provided essay specifically for the following weaknesses: {weaknesses}.
    For each weakness, clearly indicate if it is present in the essay, and if so, provide actionable and constructive criticism while referring to relevant parts of the essay. 
    After evaluating all weaknesses, assign a final score from 1 to 9, where 9 represents an exceptional essay.
    Essay: {essay}
    """

    aes_messages = [
        {"role": "system", "content": evaluator_role},
        {"role": "user", "content": full_weakness_prompt}
    ]
    
    with torch.no_grad():
        aes_outputs = pipeline(
            aes_messages,
            max_new_tokens=1024,
            eos_token_id=terminators,
            do_sample=True,
            temperature=0.6,
            top_p=0.9,
        )

    print(aes_outputs[0]["generated_text"][-1]["content"])
```
